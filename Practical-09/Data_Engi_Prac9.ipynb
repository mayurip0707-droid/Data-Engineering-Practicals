STEP 9.11 — Perform a SQL Join

```python
report = pd.read_sql(
    """
    SELECT
        o.Order_ID,
        c.Name AS Customer_Name,
        c.City,
        p.Product,
        p.Category,
        o.Quantity,
        p.Price,
        (o.Quantity * p.Price) AS Total_Amount,
        pay.Status AS Payment_Status
    FROM orders o
    JOIN customers c
        ON o.Customer_ID = c.Customer_ID
    JOIN products p
        ON o.Product_ID = p.Product_ID
    JOIN payments pay
        ON o.Order_ID = pay.Order_ID
    """,
    connection
)

display(report)
```

STEP 9.12 — Calculate Total Revenue

```python
total_revenue = pd.read_sql(
    """
    SELECT SUM(o.Quantity * p.Price) AS Total_Revenue
    FROM orders o
    JOIN products p
        ON o.Product_ID = p.Product_ID
    """,
    connection
)

display(total_revenue)
```

STEP 9.13 — Calculate Sales by Category

```python
category_report = pd.read_sql(
    """
    SELECT
        p.Category,
        SUM(o.Quantity * p.Price) AS Total_Sales
    FROM orders o
    JOIN products p
        ON o.Product_ID = p.Product_ID
    GROUP BY p.Category
    """,
    connection
)

display(category_report)
```

STEP 9.14 — Save the Report as CSV

```python
report.to_csv(
    "ecommerce_report.csv",
    index=False
)

print("E-commerce report generated successfully!")
```

```python
connection.close()

print("Database connection closed.")
```

```python
import pandas as pd
import sqlite3
import os

# Create sample e-commerce CSV files
customers_data = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5],
    "CustomerName": ["Customer A", "Customer B", "Customer C", "Customer D", "Customer E"],
    "Email": ["a@gmail.com", "b@gmail.com", "c@gmail.com", "d@gmail.com", "e@gmail.com"],
    "City": ["Mumbai", "Pune", "Nashik", "Thane", "Kalyan"]
})

products_data = pd.DataFrame({
    "ProductID": [101, 102, 103, 104, 105],
    "ProductName": ["Laptop", "Mobile", "Headphones", "Shoes", "Bag"],
    "Category": ["Electronics", "Electronics", "Electronics", "Fashion", "Fashion"],
    "Price": [55000, 25000, 2000, 3000, 2200]
})

orders_data = pd.DataFrame({
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "CustomerID": [1, 2, 3, 4, 5],
    "ProductID": [101, 102, 103, 104, 105],
    "Quantity": [1, 2, 3, 1, 2],
    "OrderDate": ["2026-01-10", "2026-02-15", "2026-03-20", "2026-04-12", "2026-05-18"]
})

payments_data = pd.DataFrame({
    "PaymentID": [501, 502, 503, 504, 505],
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "PaymentMethod": ["UPI", "Card", "Cash", "UPI", "Card"],
    "PaymentStatus": ["Paid", "Paid", "Paid", "Paid", "Pending"]
})

customers_data.to_csv("customers.csv", index=False)
products_data.to_csv("products.csv", index=False)
orders_data.to_csv("orders.csv", index=False)
payments_data.to_csv("payments.csv", index=False)

# Extract CSV data
customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")
payments = pd.read_csv("payments.csv")

# Data cleaning
customers = customers.drop_duplicates()
products = products.drop_duplicates()
orders = orders.drop_duplicates()
payments = payments.drop_duplicates()

customers["CustomerName"] = customers["CustomerName"].str.strip()
customers["Email"] = customers["Email"].str.strip().str.lower()

products["ProductName"] = products["ProductName"].str.strip()
products["Price"] = pd.to_numeric(products["Price"], errors="coerce")

orders["Quantity"] = pd.to_numeric(orders["Quantity"], errors="coerce")
orders["OrderDate"] = pd.to_datetime(orders["OrderDate"], errors="coerce")

payments["PaymentMethod"] = payments["PaymentMethod"].str.strip()
payments["PaymentStatus"] = payments["PaymentStatus"].str.strip()

# Remove invalid records
customers = customers.dropna()
products = products.dropna()
orders = orders.dropna()
payments = payments.dropna()

orders = orders[orders["Quantity"] > 0]
products = products[products["Price"] > 0]

# Create SQL database
conn = sqlite3.connect("ecommerce.db")

# Historical data loading
customers.to_sql("customers", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
orders.to_sql("orders", conn, if_exists="replace", index=False)
payments.to_sql("payments", conn, if_exists="replace", index=False)

# Newly arriving data
new_orders = pd.DataFrame({
    "OrderID": [1006, 1007],
    "CustomerID": [1, 3],
    "ProductID": [102, 101],
    "Quantity": [1, 1],
    "OrderDate": ["2026-06-10", "2026-06-15"]
})

new_orders["OrderDate"] = pd.to_datetime(new_orders["OrderDate"])
new_orders["Quantity"] = pd.to_numeric(new_orders["Quantity"])

existing_orders = pd.read_sql("SELECT OrderID FROM orders", conn)

new_orders = new_orders[
    ~new_orders["OrderID"].isin(existing_orders["OrderID"])
]

new_orders.to_sql("orders", conn, if_exists="append", index=False)

# Generate report
report = pd.read_sql("""
SELECT
    o.OrderID,
    c.CustomerName,
    p.ProductName,
    p.Category,
    p.Price,
    o.Quantity,
    (p.Price * o.Quantity) AS TotalAmount,
    o.OrderDate,
    pay.PaymentMethod,
    pay.PaymentStatus
FROM orders o
JOIN customers c ON o.CustomerID = c.CustomerID
JOIN products p ON o.ProductID = p.ProductID
LEFT JOIN payments pay ON o.OrderID = pay.OrderID
ORDER BY o.OrderDate
""", conn)

print("E-COMMERCE SALES REPORT")
print(report)

print("\nSUMMARY REPORT")
print("Total Customers:", len(customers))
print("Total Products:", len(products))
print("Total Orders:", len(report))
print("Total Sales:", report["TotalAmount"].sum())

print("\nSales by Category:")
print(
    report.groupby("Category")["TotalAmount"]
    .sum()
    .reset_index()
    .rename(columns={"TotalAmount": "TotalSales"})
)

print("\nPayment Status:")
print(report["PaymentStatus"].value_counts())

conn.close()

print("\nETL Pipeline Completed Successfully!")
```
