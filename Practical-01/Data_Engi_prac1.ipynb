## **Exercise 1 :**

```python
import pandas as pd
import re
import json
import csv
import sqlite3
import os
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup

print("All required libraries imported successfully!")
```

STEP 1 — Create a Text File

```python
text_data = """Name: Rahul
Age: 21
City: Mumbai
Course: Data Science
"""

with open("student.txt", "w") as file:
    file.write(text_data)

print("Text file created successfully!")
```

STEP 2 — Read and Parse the Text File

```python
with open("student.txt", "r") as file:
    text = file.read()

print(text)
```

STEP 3 — Extract Data from Text File

```python
student_data = {}

for line in text.splitlines():
    key, value = line.split(":", 1)
    student_data[key.strip()] = value.strip()

print(student_data)
```

STEP 4 — Create a CSV File

```python
students = [
    [1, "Rahul", 21, "Mumbai", "Data Science"],
    [2, "Sneha", 20, "Pune", "Computer Science"],
    [3, "Amit", None, "Nashik", "Data Science"],
    [4, "Priya", 22, None, "AI"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "Student_ID",
        "Name",
        "Age",
        "City",
        "Course"
    ])

    writer.writerows(students)

print("CSV file created successfully!")
```

STEP 5 — Read the CSV File

```python
df = pd.read_csv("students.csv")

display(df)
```

STEP 6 — Check Missing Values

```python
print("Missing values:")
print(df.isnull().sum())
```

STEP 7 — Check Anomalies

```python
invalid_age = df[
    (df["Age"].notnull()) &
    (df["Age"] <= 0)
]

print("Invalid age records:")
display(invalid_age)
```

```python
duplicate_ids = df[
    df["Student_ID"].duplicated()
]

print("Duplicate Student IDs:")
display(duplicate_ids)
```

STEP 8 — Create an HTML File

```python
html_data = """
<html>
<body>
<h1>Student Information</h1>

<table>
<tr>
<th>ID</th>
<th>Name</th>
<th>Course</th>
</tr>

<tr>
<td>1</td>
<td>Rahul</td>
<td>Data Science</td>
</tr>

<tr>
<td>2</td>
<td>Sneha</td>
<td>Computer Science</td>
</tr>
</table>

</body>
</html>
"""

with open("students.html", "w") as file:
    file.write(html_data)

print("HTML file created successfully!")
```

STEP 9 — Parse HTML

```python
with open("students.html", "r") as file:
    html = file.read()

soup = BeautifulSoup(html, "html.parser")

table = soup.find("table")

rows = []

for row in table.find_all("tr"):
    cells = row.find_all(["th", "td"])
    rows.append([
        cell.get_text(strip=True)
        for cell in cells
    ])

for row in rows:
    print(row)
```

STEP 10 — Create an XML File

```python
xml_data = """
<students>
    <student>
        <id>1</id>
        <name>Rahul</name>
        <course>Data Science</course>
    </student>

    <student>
        <id>2</id>
        <name>Sneha</name>
        <course>Computer Science</course>
    </student>
</students>
"""

with open("students.xml", "w") as file:
    file.write(xml_data)

print("XML file created successfully!")
```

STEP 11 — Parse XML

```python
tree = ET.parse("students.xml")
root = tree.getroot()

for student in root.findall("student"):
    student_id = student.find("id").text
    name = student.find("name").text
    course = student.find("course").text

    print(
        "ID:", student_id,
        "| Name:", name,
        "| Course:", course
    )
```

STEP 12 — Create a JSON File

```python
json_data = {
    "students": [
        {
            "id": 1,
            "name": "Rahul",
            "course": "Data Science"
        },
        {
            "id": 2,
            "name": "Sneha",
            "course": "Computer Science"
        }
    ]
}

with open("students.json", "w") as file:
    json.dump(json_data, file, indent=4)

print("JSON file created successfully!")
```

STEP 13 — Parse JSON

```python
with open("students.json", "r") as file:
    data = json.load(file)

for student in data["students"]:
    print(
        "ID:", student["id"],
        "| Name:", student["name"],
        "| Course:", student["course"]
    )
```

## **Exercise 3: Searching, Splitting and Replacing Strings using Regular Expressions**

STEP 21 — Import the re Module

```python
import re

print("Regular expression module imported successfully!")
```

STEP 22 — Searching Using Regular Expression

```python
text = "My email is rahul@gmail.com and my phone number is 9876543210."

print(text)
```

```python
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

result = re.search(pattern, text)

if result:
    print("Email found:", result.group())
else:
    print("Email not found")
```

STEP 23 — Search for a Phone Number

```python
phone_pattern = r"\b\d{10}\b"

phone = re.search(phone_pattern, text)

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found")
```

STEP 24 — Find All Matching Values

```python
text = """
Contact rahul@gmail.com
or sneha@gmail.com
or amit@yahoo.com
"""

emails = re.findall(
    r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
    text
)

print("Emails found:")
print(emails)
```

STEP 25 — Splitting Strings Using Regex

```python
text = "Apple,Orange;Banana Mango|Grapes"

print(text)
```

```python
parts = re.split(
    r"[,; |]+",
    text
)

print(parts)
```

STEP 26 — Another Splitting Example

```python
sentence = "Python is easy to learn"

words = re.split(
    r"\s+",
    sentence
)

print(words)
```

STEP 27 — Replacing Strings Using Regex

```python
text = "Contact Rahul at 9876543210."

new_text = re.sub(
    r"\b\d{10}\b",
    "XXXXXXXXXX",
    text
)

print(new_text)
```

STEP 28 — Replace Multiple Numbers

```python
text = "Rahul: 9876543210, Sneha: 9123456789"

new_text = re.sub(
    r"\b\d{10}\b",
    "XXXXXXXXXX",
    text
)

print(new_text)
```

STEP 29 — Replace Extra Spaces

```python
text = "Python     is    very      easy"

clean_text = re.sub(
    r"\s+",
    " ",
    text
)

print(clean_text)
```

## **Exercise 4: SQL CRUD Operations**

```python
import sqlite3

print("SQLite imported successfully!")
```

STEP 30 — Create the Database

```python
connection = sqlite3.connect("student_database.db")

print("Database created successfully!")
```

STEP 31 — Create a Student Table

```python
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    Student_ID INTEGER PRIMARY KEY,
    Name TEXT NOT NULL,
    Age INTEGER,
    City TEXT,
    Course TEXT
)
""")

connection.commit()

print("Students table created successfully!")
```

STEP 32 — CREATE: Insert Records

```python
students = [
    (1, "Rahul", 21, "Mumbai", "Data Science"),
    (2, "Sneha", 20, "Pune", "Computer Science"),
    (3, "Amit", 22, "Nashik", "Artificial Intelligence"),
    (4, "Priya", 21, "Thane", "Data Science")
]

cursor.executemany("""
INSERT INTO students
(Student_ID, Name, Age, City, Course)
VALUES (?, ?, ?, ?, ?)
""", students)

connection.commit()

print("Student records inserted successfully!")
```

STEP 33 — READ: Display All Records

```python
cursor.execute("SELECT * FROM students")

records = cursor.fetchall()

for record in records:
    print(record)
```

STEP 34 — READ: Select Specific Columns

```python
cursor.execute("""
SELECT Name, Course
FROM students
""")

records = cursor.fetchall()

for record in records:
    print(record)
```

STEP 35 — READ: Search for a Student

```python
cursor.execute("""
SELECT *
FROM students
WHERE Course = 'Data Science'
""")

records = cursor.fetchall()

for record in records:
    print(record)
```

STEP 36 — UPDATE: Modify a Record

```python
cursor.execute("""
UPDATE students
SET City = 'Pune'
WHERE Student_ID = 1
""")

connection.commit()

print("Student record updated successfully!")
```

STEP 37— Verify the UPDATE

```python
cursor.execute("""
SELECT *
FROM students
WHERE Student_ID = 1
""")

print(cursor.fetchone())
```

STEP 38 — DELETE: Remove a Record

```python
cursor.execute("""
DELETE FROM students
WHERE Student_ID = 4
""")

connection.commit()

print("Student record deleted successfully!")
```

STEP 39 — Verify DELETE

```python
cursor.execute("SELECT * FROM students")

records = cursor.fetchall()

for record in records:
    print(record)
```

STEP 40 — Check Number of Records

```python
cursor.execute("SELECT COUNT(*) FROM students")

count = cursor.fetchone()[0]

print("Total students:", count)
```

STEP 41 — Close the Database

```python
connection.close()

print("Database connection closed.")
```
